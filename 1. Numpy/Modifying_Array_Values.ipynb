Modifying array values 

NumPy values are mutable; so you can update the value directly

```python
import numpy as np

scores = np.array([45, 64, 32, 98, 76, 88, 94, 31, 22, 58, 99])
scores
```

```python
# Change one value : changes the score with the given index

scores[0] = 10
scores
```

```python
# Replace a slice : update multiple neighboring values at once

scores[1:3] = [20, 30]
scores 
```

```python
# Update with a condition : every score below 40 is replaced with 0

scores[scores < 40] = 0
scores
```

```python
# Update selected positions : change specific positions in once step

scores[[0, 2, 5, 8, 9]] = [10, 20, 50, 80, 90]
scores
```
