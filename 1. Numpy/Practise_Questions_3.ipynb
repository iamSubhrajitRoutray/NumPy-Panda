A teacher has 8 quiz scores and wants to practice NumPy on them.
Complete the following tasks.

1. Create a 1D NumPy array named "scores" using:
16, 25, 36, 25, 49, 64, 81, 100.

2. Update every score below 25 to 25.

3. Create a new array named "sorted_scores" by sorting the updated scores in ascending order.

4. From the updated array, find the total, average, highest value, and median.

5. Use "np.unique()" to get the distinct score values.

6. Use "np.sqrt()" to calculate the square root of each updated score.

```python
import numpy as np
```

```python
#  Q1)

scores = np.array([16, 25, 36, 25, 49, 64, 81, 100])
print(scores)
```

```python
# Q2)

updated_scores = np.where(scores < 25, 25, scores)
updated_scores
```

```python
# Q3)

sorted_scores = np.sort(updated_scores)
sorted_scores
```

```python
# Q4)

total = np.sum(updated_scores)
print(total)

average = np.mean(updated_scores)
print(average)

highest_value = np.max(updated_scores)
print(highest_value)

median_value = np.median(updated_scores)
print(median_value)
```

```python
# Q5)

unique_scores = np.unique(updated_scores, return_counts=True)
print(unique_scores)
```

```python
# Q6)

square_root = np.sqrt(updated_scores)
print(square_root)
```
