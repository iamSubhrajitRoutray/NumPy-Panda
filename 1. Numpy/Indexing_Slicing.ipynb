```python
import numpy as np

city_array = np.array(["Bengaluru", "Mumbai", "Delhi", "Bhubaneswar", "Pune", "Hyderabad", "Lucknow", "Cuttack", "Kolkota", "Jaipur", "Ahmedabad", "Chennai"])

city_grid = city_array.reshape(2, 6)
print(city_grid)
```

```python
print(city_grid[1,2])
print(city_grid[:,1:4])
print(city_grid[0,:])
print(city_grid[1,:])
print(city_grid[:,-1])
print(city_grid[:,5])
print(city_grid[1, 5::-1])
print(city_grid[::-1, -2])

```
