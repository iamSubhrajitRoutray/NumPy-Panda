```python
import numpy as np
```

```python
# Array Agrregiation Methods :  1D aggregation each method returns one value

sales = np.array([120, 200, 140, 180, 210, 170, 190])
print(sales.sum())
```

```python
# Average Sales

average = sales.mean()
print(average)

avg_round_off = np.round(average, 2)
print(avg_round_off)
```

```python
# Highest sales

print(sales.max())
```

```python
# Lowest sales

print(sales.min())
```

```python
# Median: middle value after sorting

sales.sort()
print(sales)
np.median(sales)
```

```python
# Array Agrregiation Methods :  2D aggregation: 3 stores (rows) x 4 quarters (columns)
# aixs = 0 -> down the rows, one result per column | axis = 1 -> across columns, one result per row

new_sales = np.array([
    [120, 140, 180, 200],
    [100, 170, 160, 210],
    [140, 130, 190, 220],
])

print(sales)
```

```python
# Whole array total 

print(sales.sum())
```

```python
# Column-wise sum: total of each quarter across all stores

print(sales.sum(axis=0))
```

```python
# Row-wise sum: total sale of each store

print(new_sales.sum(axis=1))
```

```python
# Column-wise average: 

new_sales.mean(axis=0)
```

```python
# Row-wise average:

new_sales.mean(axis=1)
```

```python
marks = np.array([90, 64, 90, 98, 64, 54, 54, 64, 72, 90, 100, 98])
marks
```

```python
# Unique values: unique distinct values, sorted
np.unique(marks)
```

```python
# Unique values and how many times each appears:

np.unique(marks, return_counts=True)
```

```python
#  Square root of each value:

np.sqrt(marks)
```

```python
# Position index of the highest and the lowest marks:

print(marks.argmax(), marks.argmin())
```

```python
# Running total: each value = sum of all values upto that point

print(marks.cumsum())
```

```python
# Absolute value: removes the minus sign 

profit_loss = np.array([-1, 5, -2, 4, 6, 9, 8, -3, -4, 7])
print(profit_loss)

np.abs(profit_loss)    # Removes the neagtive signs the values in the array
```
