```python
'''Q1 Create a NumPy array containing the numbers 1 to 10 and arrange it into 5 rows and 2 columns'''

import numpy as np

numbers = np.arange(1, 11)
print(numbers)

numbers.reshape(5, 2)
```

```python
'''Q2 Create another NumPy array of random decimal values between 0 and 1 in a (3 x 3) shape'''

from numpy.random import default_rng

rng = default_rng(50)

values = rng.random(9)

values.reshape(3, 3)
```
