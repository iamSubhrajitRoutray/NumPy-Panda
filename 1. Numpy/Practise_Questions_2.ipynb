A teacher has exam scores for 8 students and wants to analyze them with NumPy.
Complete the following tasks:

1. Create a 2 x 4 NumPy array called scores using these values:
35, 42, 50, 61, 28, 74, 81, 39.

2. Add 5 grace marks to every score and store the result in updated_scores.

3. From updated_scores, extract only the second row.

4. Filter the values in updated_scores that are 50 or higher.

5. Use np. where() to create a new array that shows 'Pass' for scores 50 or more and 'Needs Work' for scores below 50.

```python
import numpy as np

# Q1)

values = np.array([35, 42, 50, 61, 28, 74, 81, 39])

score = values.reshape(2, 4)

print(score)

```

```python
# Q2)

updated_score = score + 5
print(updated_score)
```

```python
#  Q3)

second_row = updated_score[1, :]
print(second_row)
```

```python
# Q4) 

print(updated_score)

criteria = updated_score > 50

print(criteria)
```

```python
# Q5)

grades = np.where(updated_score > 50, "Pass", "Needs Work")

print(grades)
```
