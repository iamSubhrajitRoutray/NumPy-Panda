```python
import numpy as np
```

```python
# ones :- Creates an array filled with 1

ones_array = np.ones((3,3), dtype = int)
ones_array
```

```python
# zeros :- Creates an array filled with 0

zeros_array = np.zeros((3,3), dtype = int)
zeros_array
```

```python
# arange :- Creates evenly stepped value in a range; np.arange(start, stop, step)

np.arange(1, 11)
```

```python
# arange :- Creates evenly stepped value in a range; np.arange(start, stop, step)

np.arange(0, 101, 10)
```

```python
# linesapce :- Creates evenly stepped value between two points linespace(start , stop, num)

np.linspace(1, 50, 10)
```

```python
# linesapce :- Creates evenly stepped value between two points with dtype as datatype = int (integer)

np.linspace(1, 50, 10, dtype = int)
```

```python
# reshape :- Changes an array into a new compatible shape reshape (row, cols)

# creating a array of (4 x 3)

temp_array = np.array(
    [[5, 4, 3],
    [7, 1, 2],
    [8, 6, 9],
    [10, 12, 11]]
)

temp_array
```

```python
# Reshape the above array of (1 x 12)

temp_array.reshape(1, 12)
```

```python
# Reshape the above array of (3 x 4)

temp_array.reshape(3, 4)
```

```python
# Reshape the above array of (2 x 6)

temp_array.reshape(2, 6)
```

```python
# Reshape the above array of (6 x 2)

temp_array.reshape(6, 2)
```
