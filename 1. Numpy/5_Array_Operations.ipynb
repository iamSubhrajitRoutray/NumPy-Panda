```python
import numpy as np

array_one = np.array([10, 20, 30, 40, 50])


array_two = np.array([1, 2, 3, 4, 5])

array_one
```

```python
# Add a Scalar : (+)

array_one += 5
array_one
```

```python
# Subtact a Scalar : (-)

array_one -= 10
array_one
```

```python
# Multiply by a Scalar : (*)

array_one *= 2
array_one
```

```python
# Divide by a Scalar : (//)

array_one //= 10
array_one
```

```python
# Element wise Addition

add_array = array_one + array_two

add_array
```

```python
# Element wise Multiplication

multiply_array = array_one * array_two

multiply_array
```
