THE WHERE FUNCTION

np.where(condition, value_if_True, value_if_False) evaluates the conditions element wise;
it returns the vvalue_if_True if it's TRUE else it returns value_if_False

```python
import numpy as np
```

```python
#  Convert Grades: Pass if greater than 50 else Fails

scores = np.array([45, 56, 76, 98, 23, 19, 44, 32, 67])
print(scores)

qualify = np.where(scores > 50, "Pass", "Fail")
print(qualify)
```

```python
#  Make number positive: replace negative with 0

nums = np.array([-4, 6, 7, -8, 2, -9, 5, 3])
print(nums)

positive = np.where(nums < 0, 0, nums)
print(positive)
```

```python
#  Categorize ages

ages = np.array([15, 56, 6, 19, 23, 9, 40, 10, 67])
print(ages)

category = np.where(ages < 18, "children", "Adult")
print(category)
```

```python
#  Apply Discounts: 10% off if price > 100

price = np.array([130, 90, 70, 50, 80, 100, 90, 430, 60, 40, 120, 110])
print(price)

discounted_price = np.where(price >= 100, price * 0.9, price)
print(discounted_price)
```

```python
#  Replace even numbers with EVEN and odd numbers with ODD

nums = np.array([15, 46, 16, 19, 22, 59, 40, 11, 66])
print(nums)

odd_even = np.where(nums % 2 == 0, "Even", "Odd")
print(odd_even)
```

PRACTISE QUESTIONS

```python
# Q1) WEATHER RPEORT: LABEL EACH DAY "HOT" IF TEMPERATURE GREATER THAN 35 ELSE "NORMAL"

temperature = np.array([23, 32, 33, 40, 21, 17, 19, 12, 34, 36, 42, 24, 14])
print(temperature)

weather = np.where(temperature > 35, "Hot", "Normal")
print(weather)
```

```python
# Q2) absent students are marked as -1; REPLACE -1 WITH 0 AND KEEP OTHER MARKS AS USUAL

nums = np.array([15, -1, 16, 19, 22, -1, 40, -1, 66])
print(nums)

odd_even = np.where(nums == -1, 0, nums)
print(odd_even)
```

```python
# Free Delivery: delivery charge is 0 if order amount is greater than or equal to 499 else 40
# Then find the total amount to pay i.e. (order price + delivery charge)

order_price = np.array([155, 556, 496, 619, 223, 1099, 240, 110, 367])
print(order_price)

total_bill = np.where(order_price < 499, order_price + 40, order_price)
print(total_bill)
```

```python
# Q4) Salary Hike: 20% hike if salary is less than 30,000 else 10% hike in salary

salary = np.array([25000, 55000, 50000, 30000, 15000, 100000, 24000, 65000, 13000])
print(salary)

hike = np.where(salary < 30000, salary * 1.2, salary * 1.1)
print(hike)
```

```python
# Q4) Phone Battery status (use 2 arrays together):
#  "Charging" -> if the phone is plugged in
#  "Charging Now" -> if not plugged and the battery is less than 20
#  "Ok" -> otherwise

battery = np.array([15, 29, 45, 7, 19, 50, 24, 9])
print(battery)

is_charging = np.array([False, True, False, False, True, False, True, False])
print(is_charging)

status = np.where (is_charging == True, "Charging", np.where(battery < 20, "Charge Now", "Ok"))
print(status)
```

```python
# Q5) (Challenge) Stock report using yesterday's and today's price:

# First find the percentage(%) change = ((today - yesterday) / yesterday * 100)

# Then label:
# "Big Jump" if change > 5,
# "Up" if change > 0,
# "Down" if change < 0, else "No change"

yesterday = np.array([200, 150, 320, 90, 500, 75])
print(yesterday)

today = np.array([215, 150, 310, 93, 500, 72])
print(today)

percentage_change = ((today - yesterday)/yesterday * 100)
print(percentage_change)

change = np.where(percentage_change > 5, "Big Jump", np.where(percentage_change > 0, "Up", "Down"))
print(change)

```
