```python
import numpy as np
```

```python

# Original 3 x 3 array:

test_array = np.array([
    [1, 2, 3],
    [1, 2, 3],
    [1, 2, 3],
])
test_array
```

```python
# scalar: 1 is added to every array

i1 = test_array + 1
print(i1)
```

```python
# row vector (3,): stretch down the rows

r1 = test_array + np.array([3,2,1])
print(r1)
```

```python
# column vector (3,1): stretch across the columns

c1 = test_array + np.array([[3],
                            [2],
                            [1]])
print(c1)
```

```python
# row vector + column vector : both stretch to (3, 3)

rc1 = np.array([1, 2, 3]) + np.array([[3],
                                    [2],
                                    [1]])
print(rc1)
```
