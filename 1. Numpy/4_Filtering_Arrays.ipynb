Filtering Arrays

It creats a condition (True or False) for each element. NumPy returns only the elements where the condition is True.

```python
import numpy as np

ages = np.array([4, 9, 13, 15, 14, 23, 45, 32, 10, 24, 19, 22, 34, 31, 8, 6])

ages
```

```python
# Filering = array[condition] : True values are kept; False values are ignored

adult_mask = ages > 18

adult_mask
```

```python
adult_mask = ages[ages > 18]

adult_mask
```

```python
children_mask = ages[ages < 18]

children_mask
```

```python
adults = ages > 18

ages[adults]
```

```python
children = ages < 18

ages[children]
```

```python
#  "&" means AND operation; both conditions must be True 

teens = (ages >= 14) & (ages <= 18)

ages[teens]
```

```python
#  "&" means AND operation; both conditions must be True

working_adults = (ages >= 20) & (ages <= 60)

ages[working_adults]
```

```python
#  "|" means OR operation; either can be True

speical_mask = (ages < 10) | (ages > 60)

ages[speical_mask]
```
