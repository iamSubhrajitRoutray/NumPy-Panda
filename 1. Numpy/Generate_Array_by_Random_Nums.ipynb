```python
'''1. default_rng : Creates a random number geenrator object,
Add a seed if you want the same numbers again later'''

from numpy.random import default_rng
rng = default_rng(100)                # default_rng(seed)
rng.random(5)

```

```python
'''2. random : Returns random decimal numbers between 0 and 1'''

values = rng.random(10)
values
```

```python
random_nums = rng.integers(low=1, high=100, size=5)
random_nums
```

```python
'''normal : retuns random numbers around the center value; valeus tends to cluster near the mean'''

marks = rng.random(10)
marks
```

```python
rng.normal(loc=70, scale= 10, size=10)
```
